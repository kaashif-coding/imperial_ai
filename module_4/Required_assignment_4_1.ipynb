# Required assignment 4.1 Calculating MLE in Python

Maximum likelihood estimation (MLE) is a statistical method used to estimate the parameters of a probability distribution or a statistical model based on observed data. The goal of MLE is to find the values of the model parameters that maximise the likelihood of observing the data given in the model. This method is widely used in various fields, such as economics, finance, biology and engineering, for analysing relationships between variables and making predictions.


The steps to implement MLE are:

- Assume a probability distribution.

- Define the likelihood function.

- Maximise the log likelihood.

- Compute the maximum likelihood estimate.

## Question 1: Convert `sample` to a binary array.

```python
###GRADED
n = len(sample)
heads_total = sum(sample)

print("The number of coin tosses is given by",n)
print("The number of heads in total is given by",heads_total)
```

- What is the maximum number of consecutive heads appearing together? Store this to a variable named `max_running`.


```python
### GRADED
running_heads = 0
max_running = 0

for toss in sample:
    if toss == 1:  # If it's a head
        running_heads += 1
        max_running = max(max_running, running_heads)  # Update max_running if current run is longer
    else:
        running_heads = 0  # Reset the count for tails

print("The maximum number of consecutive heads is",max_running)
```

## Question 3: Find the MLE estimate for $\theta$.

- $\theta$ is the probability of head showing in a single toss. 

- Compute $\hat{\theta}$.

```python
### GRADED
theta_hat = np.mean(sample)

print("The MLE estimate of theta is", theta_hat)
```

## Question 5: After how many tosses does your MLE estimate falls within the range of $[0.96999, 0.97001]$? 

- Report the minimum number of tosses required for the MLE estimate to fall within this range.

- To compute this, first calculate the ratios, i.e. the cumulative sum of elements, in `sample_2` divided by the number of trials at each step.

- Assign the `truth` value between the range of $[0.96999, 0.97001]$.

- Compute the `smallest-accurate-n`.

```python
### GRADED
ratios = np.cumsum(sample_2) / np.arange(1, len(sample_2) + 1)
truth = (ratios >= 0.96999) & (ratios <= 0.97001)
smallest_accurate_n = np.where(truth)[0][0] + 1

print("The minimum number of tosses is", smallest_accurate_n)
```
