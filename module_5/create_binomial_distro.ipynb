```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

n = 10
p = 0.6

s = np.arange(0, n + 1)

# Apply the binomial formula to each count:  comb(n,s) * p^s * (1-p)^(n-s)
probs = np.array([comb(n, k) * (p**k) * ((1 - p)**(n - k)) for k in s])

print("Probabilities sum to:", probs.sum())   # sanity check → 1.0

plt.bar(s, probs, color='#185FA5')
plt.xlabel('Number of successes (heads)')
plt.ylabel('Probability')
plt.title(f'Binomial distribution (n={n}, p={p})')
plt.xticks(s)
plt.show()
```
