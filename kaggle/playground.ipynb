```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

data = {
    'X1': [1, 2, 3, 4, 5],
    'X2': [5, 7, np.nan, 11, 13],
    'Y': [10, 14, np.nan, 22, 26]
}
df = pd.DataFrame(data)

# Impute missing values in X2 with mean
mean_x2 = df['X2'].dropna().mean()
df['X2'] = df['X2'].fillna(mean_x2)  # avoid inplace on a slice

# Separate rows — use .copy() to avoid SettingWithCopyWarning
train_data = df[df['Y'].notnull()].copy()
test_data = df[df['Y'].isnull()].copy()

# Fit model
model = LinearRegression()
model.fit(train_data[['X1', 'X2']], train_data['Y'])

# Predict missing Y — now safe because test_data is a proper copy
test_data['Y'] = model.predict(test_data[['X1', 'X2']])

# Combine
df_imputed = pd.concat([train_data, test_data]).sort_index()
print(df_imputed)
```
