# Required assignment 2.1: Dealing with missing data


Missing data, also known as missing values, refers to the absence of data entries in a data set when values are expected but not recorded. These gaps can occur for various reasons, such as incomplete data entry, equipment failures, lost files or survey non-response. Missing data can significantly impact research integrity and business decisions by skewing results and reducing the sample size that is available for analysis.

### Filling in missing data using `fillna()`

The `dropna()` function removes 13 rows with missing values. However, dropping data isn’t always recommended, especially if it leads to significant information loss. A more common alternative is to fill in missing values using the mean, median or mode, depending on the nature of the data.


### Question 4: Calculate the mean and median for imputation.


- Before imputing missing values, calculate the mean and median of the Apps and Accept columns from the original df_college DataFrame.
- Store these calculated values in the variables `mean_of_apps`,`mean_of_accepts`,`median_of_apps` and `median_of_accepts`, respectively.
- You will use these values in the subsequent imputation steps.


```python
###GRADED
mean_of_apps = df_college['Apps'].mean()
mean_of_accepts = df_college['Accept'].mean()
median_of_accepts = df_college['Accept'].median()
median_of_apps = df_college['Apps'].median()

#Make a copy of df_college to fill in the missing data
df_college_v1 = df_college.copy(deep = True)
```

### Question 5: Impute the missing values using mean.


- Utilising the previously calculated mean values (`mean_of_apps` and `mean_of_accepts`), fill in the missing entries in the `Apps` and `Accept` columns of the `df_college_v1` DataFrame.
- Ensure that these changes are applied directly to the `df_college_v1` DataFrame.

```python
###GRADED
#fill in the missing values in the copy df_college_v1
df_college_v1['Apps'] = df_college_v1['Apps'].fillna(mean_of_apps)
df_college_v1['Accept'] = df_college_v1['Accept'].fillna(mean_of_accepts)
```

### Question 6: Impute the missing values using median.


- Create a deep copy of the original df_college DataFrame and name it `df_college_v2`.
- Subsequently, fill in the missing values in the `Apps` and `Accept` columns of `df_college_v2` with their respective median values (`median_of_apps` and `median_of_accepts`), which you calculated earlier.

```python
###GRADED
# Fill in the missing values in the copy df_college_v2
df_college_v2 = df_college.copy(deep = True)
df_college_v2['Apps'] = df_college_v2['Apps'].fillna(median_of_apps)
df_college_v2['Accept'] = df_college_v2['Accept'].fillna(median_of_accepts)
```

### Question 7: Impute the missing values using mode.

- Create a deep copy of the original df_college DataFrame and name it `df_college_v3`.
- Find the mode for the Apps and Accept columns and store the respective variable `mode_of_apps`, `mode_of_accepts`.
- Take the first mode value. Hint: `df_college["Apps"].mode()[0]`
- Subsequently, fill in the missing values in the Apps and Accept columns of `df_college_v3` with their respective mode values.

```python
###GRADED
mode_of_apps = None
mode_of_accepts = None
df_college_v3 = df_college.copy(deep = True)


# YOUR CODE HERE
mode_of_apps = df_college['Apps'].mode()[0]
mode_of_accepts = df_college['Accept'].mode()[0]

df_college_v3['Apps'] = df_college_v3['Apps'].fillna(mode_of_apps)
df_college_v3['Accept'] = df_college_v3['Accept'].fillna(mode_of_accepts)
```

### Filling in the missing rows using regression

You can impute missing values using a regression approach. The steps to implement this method are:

- Separate rows with and without missing values in `Apps`.

- Define the predictors and target.

- Fit a linear regression model.

- Predict the missing `Apps` values.

- Replace the missing `Apps` values with the predictions.

```python
df_college.isna().sum()
```

```python
df_college_selected=df_college[['Apps', 'Accept']]
print(df_college_selected.isna().sum())
```

```python
#The Apps and Accept columns have null values
#Therefore, only those those columns are selected
#For regression, you need to separate rows with and without missing rows for training and testing data sets
#Separate rows with and without missing values in the Apps column
train_apps = df_college[df_college['Apps'].notnull()]
test_apps = df_college[df_college['Apps'].isnull()]
```

```python
#Define the predictors and target
X_train_apps = train_apps[['Enroll', 'Top10perc', 'Top25perc', 'F.Undergrad', 'P.Undergrad', 'PhD', 'S.F.Ratio', 'Grad.Rate']]
y_train_apps = train_apps['Apps']
```

```python
#Fit a linear regression model
model_apps = LinearRegression()
model_apps.fit(X_train_apps, y_train_apps)
```

```python
#Predict the missing Apps values
X_test_apps = test_apps[['Enroll', 'Top10perc', 'Top25perc', 'F.Undergrad', 'P.Undergrad', 'PhD', 'S.F.Ratio', 'Grad.Rate']]
predicted_apps = model_apps.predict(X_test_apps)
```

### 

```python
#Replace the missing Apps values with predictions
df_college.loc[df_college['Apps'].isnull(), 'Apps'] = predicted_apps
```
